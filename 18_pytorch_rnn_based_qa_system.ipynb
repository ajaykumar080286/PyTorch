```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/PyTorch/main/100_Unique_QA_Dataset.csv")
```

```python
df.head()
```

```python
from transformers import AutoTokenizer

# Choose a pre-trained tokenizer. 'bert-base-uncased' is a general-purpose BERT tokenizer.
# You can replace it with other models like 'roberta-base', 'distilbert-base-uncased', etc.
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')

# Example text to tokenize: using the first question from your DataFrame
sample_text = df['question'].iloc[0]
print(f"Sample Text: {sample_text}")

# Tokenize the text.
# `padding=True` pads shorter sequences to the maximum length.
# `truncation=True` truncates longer sequences.
# `return_tensors="pt"` returns PyTorch tensors (can be "tf" for TensorFlow or None for Python lists).
tokenized_output = tokenizer(sample_text, padding=True, truncation=True, return_tensors="pt")

print("\nTokenized Output (input_ids):")
print(tokenized_output['input_ids'])

print("\nDecoded Tokens:")
# Decode the token IDs back to human-readable tokens
print(tokenizer.convert_ids_to_tokens(tokenized_output['input_ids'][0]))

print("\nAttention Mask:")
print(tokenized_output['attention_mask'])

# If you have multiple texts, you can tokenize them as a batch:
# batch_texts = df['question'].tolist()[:3] # Taking first 3 questions as an example
# batch_tokenized_output = tokenizer(batch_texts, padding=True, truncation=True, return_tensors="pt")
# print("\nBatch Tokenized Output (first 3 questions):")
# print(batch_tokenized_output['input_ids'])
```
